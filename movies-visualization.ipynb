```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt 
```

```python
df = pd.read_csv("movies_updated.csv")
df.head()
```

```python
for col in ['rating', 'writer', 'star', 'company']:
    df[col] = df[col].fillna('Unknown')

df['gross'] = df['gross'].fillna(df['gross'].median())
df.isnull().sum()

df.head() 
df.info()
df.describe()
```

```python
len(df) 

num_cols = df.select_dtypes(include='number').columns
print(num_cols)

cat_cols = df.select_dtypes(include=['object']).columns
print(cat_cols)
```

```python
# number of movies released each year.
df['year'].unique()
year_value = df['year'].value_counts().sort_index()
print(year_value)
```

```python
plt.plot(year_value.index, year_value.values,marker = '*')
plt.xlabel("Year")
plt.ylabel("Number of Movies")
plt.title("Movies Released Per Year")
plt.show()
```

```python
# top 10 genera
top_10_genres = df['genre'].value_counts().head(10)
print(top_10_genres)
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.bar(top_10_genres.index, top_10_genres.values)

ax.set_title("Top 10 Genres with the Largest Number of Movies")
ax.set_xlabel("Genre")
ax.set_ylabel("Number of Movies")

plt.xticks(rotation=45)

plt.show()
```

```python
#  top directories
top_10_directors = df['director'].value_counts().head(10)
print(top_10_directors)

fig, ax = plt.subplots(figsize=(10, 6))


ax.barh(top_10_directors.index, top_10_directors.values)

ax.set_title("Top 10 Directors with the Largest Number of Movies")
ax.set_xlabel("Number of Movies")
ax.set_ylabel("Director")


ax.invert_yaxis()
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.hist(df["score"], bins=10)

ax.set_title("Distribution of Movie Scores")
ax.set_xlabel("Score")
ax.set_ylabel("Frequency")

plt.show()
```

```python
# Group by genre and calculate mean score
genre_avg = df.groupby("genre")["score"].mean()

# Sort from highest to lowest
genre_avg = genre_avg.sort_values(ascending=False)
print(genre_avg)
```

```python
fig, ax = plt.subplots(figsize=(10, 6))

ax.bar(genre_avg.index, genre_avg.values)

ax.set_title("Average Movie Score by Genre")
ax.set_xlabel("Genre")
ax.set_ylabel("Average Score")

plt.xticks(rotation=45)

plt.show()
```

```python
# Top 10 movies by gross revenue
top_10 = df.sort_values("gross", ascending=False).head(10)

fig, ax = plt.subplots(figsize=(10, 6))

ax.barh(top_10["name"], top_10["gross"])

ax.set_title("Top 10 Movies by Gross Revenue")
ax.set_xlabel("Gross Revenue")
ax.set_ylabel("Movie Name")

ax.invert_yaxis()

plt.show()
```
